## 2. Locate the project

This cell handles three cases automatically:
- You uploaded/extracted the `vajra-ai-phase-ai-00` folder directly into the Colab file browser (`/content/vajra-ai-phase-ai-00`).
- You uploaded `vajra-ai-phase-ai-00.zip` into `/content/` and it still needs unzipping.
- You put either of the above inside Google Drive — edit `DRIVE_PATH` below if so.

Edit `DRIVE_PATH` only if your folder/zip lives somewhere specific in Drive; otherwise leave it as-is.

```python
import os
import zipfile
import glob

PROJECT_NAME = 'vajra-ai-phase-ai-00'
DRIVE_PATH = '/content/drive/MyDrive'  # edit if your project is elsewhere in Drive

candidates = [
    f'/content/{PROJECT_NAME}',
    f'{DRIVE_PATH}/{PROJECT_NAME}',
]

project_root = None
for c in candidates:
    if os.path.isdir(c):
        project_root = c
        break

if project_root is None:
    # Look for a zip to extract, in /content or Drive
    zip_candidates = glob.glob('/content/*.zip') + glob.glob(f'{DRIVE_PATH}/*.zip')
    zip_path = None
    for z in zip_candidates:
        if PROJECT_NAME in os.path.basename(z):
            zip_path = z
            break
    if zip_path is not None:
        print(f'Found zip: {zip_path} — extracting to /content/ ...')
        with zipfile.ZipFile(zip_path, 'r') as zf:
            zf.extractall('/content/')
        project_root = f'/content/{PROJECT_NAME}'

if project_root is None or not os.path.isdir(project_root):
    raise FileNotFoundError(
        f"Could not find '{PROJECT_NAME}' folder or a matching zip in /content or Drive. "
        "Upload the folder or zip, then re-run this cell."
    )

os.chdir(project_root)
print(f'Working directory set to: {os.getcwd()}')
```
